# pipeline walkthrough

compile the pipeline locally and inspect the DAG before submitting it to KFP.

```python
from kfp.v2 import compiler
from pipelines.demand_forecast import demand_forecast_pipeline
compiler.Compiler().compile(demand_forecast_pipeline, 'dist/demand_forecast.json')
print('ok')
```

```python
import json, pathlib
spec = json.loads(pathlib.Path('dist/demand_forecast.json').read_text())
print('components:', list(spec.get('pipelineSpec', {}).get('components', {}).keys()))
```

```python
# submit
import kfp, os
client = kfp.Client(host=os.environ['KFP_HOST'])
exp = client.create_experiment(name='demand-forecast')
run = client.run_pipeline(
    experiment_id=exp.id,
    job_name='demand-forecast-manual',
    pipeline_package_path='dist/demand_forecast.json',
    params={'project': os.environ['GCP_PROJECT']},
)
run.id
```
